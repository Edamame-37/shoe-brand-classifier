# Setup dependencies

Install dependency langsung ke kernel Jupyter aktif tanpa virtual environment.

```python
import sys

python_version = sys.version_info
print(f"Python executable: {sys.executable}")
print(f"Python version: {python_version.major}.{python_version.minor}.{python_version.micro}")

if not ((3, 10) <= python_version[:2] < (3, 14)):
    raise RuntimeError("Gunakan kernel Python >=3.10 dan <3.14 untuk inference-sdk.")
```
